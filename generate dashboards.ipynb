```python
%sql
SELECT order_approved_at::date AS date,
       COUNT(DISTINCT seller_id) AS active_sellers
FROM olist_eccom_catalog.gold.fact_orders
GROUP BY date
ORDER BY date;
```

```python
%sql
SELECT COUNT(*) AS total_customers
FROM olist_eccom_catalog.gold.dim_customers;
```

```python
%sql
SELECT customer_state, COUNT(*) AS customer_count
FROM olist_eccom_catalog.gold.dim_customers
GROUP BY customer_state
ORDER BY customer_count DESC;
```

```python
%sql
SELECT 
    CASE WHEN order_count > 1 THEN 'Repeat Customer'
         ELSE 'New Customer' END AS customer_type,
    COUNT(*) AS count
FROM (
    SELECT customer_id, COUNT(*) AS order_count
    FROM olist_eccom_catalog.gold.fact_orders
    GROUP BY customer_id
)
GROUP BY customer_type;
```
