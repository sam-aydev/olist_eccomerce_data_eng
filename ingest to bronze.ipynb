```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
%sql
CREATE CATALOG IF NOT EXISTS olist_eccom_catalog
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS olist_eccom_catalog.bronze
```

```python
%sql
CREATE VOLUME IF NOT EXISTS olist_eccom_catalog.bronze.raw;
```

### My Parameters 

```python
file_name="olist_geolocation_dataset.csv" 


table_name="geolocation_bronze"
```

```python
df = spark.read.option("header", "true").option("inferSchema", "true").csv(f"/Volumes/olist_eccom_catalog/bronze/raw/{file_name}")
df.display()
```

```python
df.write.format("delta").mode("overwrite").saveAsTable(f"olist_eccom_catalog.bronze.{table_name}")
```
